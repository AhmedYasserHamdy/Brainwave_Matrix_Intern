## Data Cleaning

```python
# Convert 'Date' to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Extract day, month, and weekday
df['Day'] = df['Date'].dt.day
df['Month'] = df['Date'].dt.month
df['Weekday'] = df['Date'].dt.day_name()

# Convert 'Time' to datetime and extract hour
df['Time'] = pd.to_datetime(df['Time'])
df['Hour'] = df['Time'].dt.hour
```

```python
df.dtypes
```

## Data Analysis 

```python
# Analyze the relationship between sales and customer segments
segment_sales = df.groupby(['Gender', 'Customer type', 'City'])['Sales'].mean()
print("Average Sales by Customer Segment:\n", segment_sales)
```

```python
# Investigate the impact of payment method on sales
payment_sales = df.groupby('Payment')['Sales'].mean()
print("\nAverage Sales by Payment Method:\n", payment_sales)
```

```python
# Analyze the relationship between product line and sales
product_sales = df.groupby('Product line')['Sales'].sum()  # Use sum for total revenue
print("\nTotal Revenue by Product Line:\n", product_sales)
```

```python
# Analyze the relationship between product line and sales
product_sales = df.groupby('Product line')['Sales'].mean()  # Use mean for avg revenue
print("\nTotal Revenue by Product Line:\n", product_sales)
```

```python
# Explore the relationship between customer ratings and various factors
rating_product = df.groupby('Product line')['Rating'].mean()
print("\nAverage Rating by Product Line:\n", rating_product)
rating_payment = df.groupby('Payment')['Rating'].mean()
print("\nAverage Rating by Payment Method:\n", rating_payment)
rating_city = df.groupby('City')['Rating'].mean()
print("\nAverage Rating by City:\n", rating_city)
```

```python
# Calculate the total revenue for each branch
branch_revenue = df.groupby('Branch')['Sales'].sum()
print("\nTotal Revenue by Branch:\n", branch_revenue)
```

```python
# Group by hour and calculate average sales
hourly_sales = df.groupby('Hour')['Sales'].mean()
print("\nAverage Sales by Hour of Day:\n", hourly_sales)
```

## Data Visualization

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='Product line', y='Sales', data=df, estimator='mean', ci=None)
plt.title('Average Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Average Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
# This line creates a bar chart showing the average sales for each product line.
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='Product line', y='Rating', data=df, estimator='mean', ci=None)
plt.title('Average Customer Rating by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Average Rating')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
# this one shows the average customer rating for each product line.
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Payment', data=df)
plt.title('Distribution of Sales across Payment Methods')
plt.xlabel('Payment Method')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
#  count plot showing the distribution of sales across different payment methods. It displays the number of transactions for each payment method.
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Hour', y='Sales', data=df, estimator='mean', ci=None)
plt.title('Average Sales by Hour of the Day')
plt.xlabel('Hour')
plt.ylabel('Average Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
# This bar chart illustrates the average sales for each hour of the day. This helps visualize potential peak sales hours.
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Rating', y='Sales', data=df)
plt.title('Relationship between Customer Ratings and Sales')
plt.xlabel('Customer Rating')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
# This scatter plot explores the relationship between customer ratings and sales. It plots individual data points, allowing you to visually assess if there's a correlation between these two variables
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x='Branch', y='Sales', data=df, estimator=sum, ci=None)
plt.title('Total Revenue for Each Branch')
plt.xlabel('Branch')
plt.ylabel('Total Revenue')
plt.tight_layout()
plt.show()
# bar chart shows the total revenue generated by each branch of the supermarket.
```

```python
plt.figure(figsize=(16, 6))
sns.barplot(x='City', y='Sales', hue='Gender', data=df, estimator='mean', ci=None)
plt.title('Average Sales by Customer Segment (City and Gender)')
plt.xlabel('City')
plt.ylabel('Average Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
# bar chart displays the average sales by city, further broken down by gender. The hue='Gender' argument creates separate bars for each gender within each city.
```

```python
plt.figure(figsize=(16, 6))
sns.barplot(x='City', y='Sales', hue='Customer type', data=df, estimator='mean', ci=None)
plt.title('Average Sales by Customer Segment (City and Customer Type)')
plt.xlabel('City')
plt.ylabel('Average Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
# this one shows the average sales by city, but this time it's broken down by customer type (member or normal).
```

## Purpose of the Visualizations

### The overall goal of this visualization code is to create a visual representation of the data to make it easier to understand trends, patterns, and relationships within the supermarket sales data. These visualizations can be used to:

#### Identify top-performing product lines.
#### Understand customer preferences for payment methods.
#### Determine peak sales hours.
#### Explore the relationship between customer ratings and sales.
#### Compare the performance of different branches.
#### Analyze sales trends across different customer segments.
